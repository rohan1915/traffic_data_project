####Saving the external locations in variables

```python

bronze_path = spark.sql('describe external location `bronze`').select('url').collect()[0][0]
silver_path = spark.sql('describe external location `silver`').select('url').collect()[0][0]
gold_path = spark.sql('describe external location `gold`').select('url').collect()[0][0]
```

####Passing the environments dynamically

```python
dbutils.widgets.text(name='env',defaultValue='',label='Enter the Environment')
env = dbutils.widgets.get('env')
```

####Defining functions to use catalog and create schemas dynamically

```python
def create_bronze_schema(environment,path,schema_name):
    print(f'Using {environment}catalog')
    spark.sql(f"USE CATALOG '{environment}catalog'")
    print(f"Creating {schema_name} schema in {path}")
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{schema_name}` MANAGED LOCATION '{path}/{schema_name}'")
    print('****************************')
```

```python
def create_silver_schema(environment,path,schema_name):
    print(f'Using {environment}catalog')
    spark.sql(f"USE CATALOG '{env}catalog'")
    print(f"Creating {schema_name} schema in {path}")
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{schema_name}` MANAGED LOCATION '{path}/{schema_name}'")
    print('****************************')
```

```python
def create_gold_schema(environment,path,schema_name):
    print(f'Using {environment}catalog')
    spark.sql(f"USE CATALOG '{env}catalog'")
    print(f"Creating {schema_name} schema in {path}")
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{schema_name}` MANAGED LOCATION '{path}/{schema_name}'")
    print('****************************')
```

### Creating the tables

```python
def raw_traffic_table(environment):
    print(f"Creating raw_traffic table in {environment} catalog")
    spark.sql(f"""DROP TABLE IF EXISTS `{environment}catalog`.`bronze`.`raw_traffic`""")
    spark.sql(f"""CREATE TABLE IF NOT EXISTS `{environment}catalog`.`bronze`.`raw_traffic`
              (
                  Record_ID INT,
                  Count_point_id INT,
                  Direction_of_travel VARCHAR(255),
                  Year INT,
                  Count_date VARCHAR(255),
                  hour INT,
                  Region_id INT,
                  Region_name VARCHAR(255),
                  Local_authority_name VARCHAR(255),
                  Road_name VARCHAR(255),
                  Road_category_ID INT,
                  Start_junction_road_name VARCHAR(255),
                  End_junction_road_name VARCHAR(255),
                  Latitude DOUBLE,
                  Longitude DOUBLE,
                  Link_length_km DOUBLE,
                  Pedal_cycles INT,
                  Two_wheeled_motor_vehicles INT,
                  Cars_and_taxis INT,
                  Buses_and_coaches INT,
                  LGV_type INT,
                  HGV_type INT,
                  EV_car INT,
                  EV_bike INT,
                  Extract_time TIMESTAMP
              )""")
    
    print('****************************')
```

```python
def raw_road_table(environment):
    print(f"Creating raw_roads table in {environment} catalog")
    spark.sql(f"""DROP TABLE IF EXISTS `{environment}catalog`.`bronze`.`raw_roads`""")
    spark.sql(f"""CREATE TABLE IF NOT EXISTS `{environment}catalog`.`bronze`.`raw_roads`
              (
                  Road_id INT,
                  Road_category_id INT,
                  Road_category VARCHAR(20),
                  Region_id INT,
                  Region_name VARCHAR(255),
                  Total_link_length_km DOUBLE,
                  Total_link_length_miles DOUBLE,
                  All_motor_vehicles DOUBLE,
                  Extract_time TIMESTAMP
              )""")
    
    print('****************************')
```

####Calling the functions

```python
create_bronze_schema(env,bronze_path,'bronze')
create_silver_schema(env,silver_path,'silver')
create_gold_schema(env,gold_path,'gold')

raw_traffic_table(env)
raw_road_table(env)
```
